```python
"""
Solution for week 4 practice of Data Visualization

Animate sine wave using FuncAnimation from matplotlib.animation
"""

import numpy as np
from matplotlib import pyplot as plt
from matplotlib import animation
```

```python
def animate_sine(file_name=None):
    """
    Input: optional string file_name
    
    Action: Plot animated sine wave. Save result in specified file_name if present.
    """
    
    def animate(frame_num):
        """
        Input: Integer frame_num
        
        Action: Updates line with data for given frame
        """
        xvals = np.linspace(0, 2, 201)
        yvals = np.sin(2 * np.pi * (xvals - 0.01 * frame_num))
        line.set_data(xvals, yvals)
        return (line,)
    
    fig, axs = plt.subplots()
    axs.set_xlim(0, 2)
    axs.set_ylim(-2, 2)
    line, = axs.plot([], [], "r", lw=2)
    
    anim = animation.FuncAnimation(fig, animate, init_func=lambda : animate(0),
                                   frames=200, interval=20, blit=True)
    
    # Can only save animation in Vocareum where proper encoders are installed
    if file_name:
        anim.save(file_name, fps=30, extra_args=['-vcodec', 'libx264'])
        
    return anim
```
