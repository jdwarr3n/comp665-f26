```python
import pandas as pd
```

## Create Series 
A series is a one-dimensional array-like object. It labels each data item. The associated array of data labels is called the index of series. 

```python
# Create Series with default index:
ser1 = pd.Series([5, 8, -3])
print(ser1)
```

```python
# Create Series with assigned index:
ser2 = pd.Series([5, 8, -3], index=["a", "b", "c"])
print(ser2)
```

### Selection on Series

Passing **an index** - either the default or the assigned index

```python
ser2['a']
```

```python
ser2.iloc[0]
```

Passing **a list of indices** - either the default or the assigned indices

```python
ser2[['a', 'c']]
```

```python
ser2.iloc[[0, 2]]
```

Passing **a slice of default indices** 

```python
ser2[:2]
```

## Create DataFrames 
A dataframe is a 2-dimentional data structure. It represents a rectangular table of data with rows and columns. Columns can be of different data types. Just like series, dataframes label the data.

```python
# Create a dataframe from dictionary of lists:
dict_of_lists = {"name" : ["Nick", "Tom", "Rose"], 
                 "title" : ["Professor", "Attorney", "Physician"], 
                 "age" : [58, 45, 36]}

df1 = pd.DataFrame(dict_of_lists)
print(df1)
```

```python
df1 = pd.DataFrame(dict_of_lists, index=["p1", "p2", "p3"])
print(df1)
```

```python
# Create a dataframe from list of lists:
list_of_lists = [["New York City", 8601186, 2019], ["Los Angeles", 4057841, 2019], ["Chicago", 2679044, 2019]] 
df2 = pd.DataFrame(list_of_lists)
print(df2)
```

```python
df2 = pd.DataFrame(list_of_lists, columns=["city", "population", "year"])
print(df2)
```

### Selection on DataFrames

#### Selecting columns using standard Python dictionary notation

```python
df1['age']
```

#### Selecting rows and columns using `loc` and `iloc` methods
`loc` and `iloc` methods are used to select from rows and columns.  
`loc` method gets the rows/columns labels (assigned indices). 
`iloc` method gets the rows/columns positions starting from zero (default indices). 

Selecting just rows using `iloc` method

```python
df1.iloc[0]
```

```python
df1.iloc[[0, 2]]
```

```python
df1.iloc[:2]
```

Selecting both rows and columns using `iloc` method

```python
df1.iloc[0, 2]
```

```python
df1.iloc[[0, 2], [1, 2]]
```

```python
df1.iloc[:, :2]
```
