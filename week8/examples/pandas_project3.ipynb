```python
"""
Reading, cleaning, and plotting financial time series from stock indices
Rework of Project 3 using pandas

Dates - https://stackoverflow.com/questions/51208817
Cleaning - https://stackoverflow.com/questions/40011531
"""

import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Financial data files
DATA_PATH = "../data/"

DJIA = "DJIA.csv"
NASDAQ = "NASDAQ.csv"
SP500 = "SP500.csv"
```

```python
def plot_index(file_name):
    """
    Input: String file_name for a CSV file with financial data
    Action: Reads, cleans, and plots the data using dates as the x-axis
    """
    
    raw_df = pd.read_csv(file_name, index_col="DATE", parse_dates=True)
    price_col = raw_df.iloc[:, 0]
    cleaned_series = pd.to_numeric(price_col, errors="coerce")
    cleaned_series.plot(legend=True)  
```

```python
def test_plot_index():
    
    plot_index(DATA_PATH + DJIA)
    plot_index(DATA_PATH + NASDAQ)
    plot_index(DATA_PATH + SP500)
    plt.title("Price for three major stock indices over the last decade")
    
test_plot_index()
```
