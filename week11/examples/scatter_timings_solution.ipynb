```python
"""
Solution for week 11 practice in Data Visualization

Comparing two method for generating a scatter plot from a numpy array of points
Note that you should use the pyplot API for this practice project
"""

import time
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Define a numpy array of random points

def gen_random_points(num_points):
    """
    Input: integer num_points
    
    Output: numpy array of points with random (x, y) coordinates of length num_pts
    """
    
    return np.random.rand(num_points, 2)
```

```python
# Implement two plotting methods

def plot_points_individually(points, title="Plot of points using one call to scatter() per point"):
    """
    Input: numpy array points of (x, y) coordinate pairs, optional string title
    
    Action: Plot points using one call to scatter() per point
    """

    for point in points:
        plt.scatter(point[0], point[1])
    plt.title(title)
    #plt.show()
    plt.clf()
    
plot_points_individually(gen_random_points(1000))
```

```python
def plot_points_together(points, title="Plot of points using a single call to scatter()"):
    """
    Input: numpy array points of (x, y) coordinate pairs, optional string title
    
    Action: Plot points using a single call to scatter()
    """

    plt.scatter(points[:, 0], points[:, 1])
    plt.title(title)
    #plt.show()
    plt.clf()
    
plot_points_together(gen_random_points(1000))
```

```python
# Replace show() by clf() to clear figures between calls to plotting functions

def compare_plot_points(point_range):
    """
    Input: range object point_range
    
    Actiom: Plot comparison of running times two versions of point plotters
    """
    
    individual = []
    together = []
    for num_points in point_range:
        points = gen_random_points(num_points)

        begin_time = time.perf_counter()
        plot_points_individually(points)
        end_time = time.perf_counter()
        individual.append((num_points, end_time - begin_time))
    
        begin_time = time.perf_counter()
        plot_points_together(points)
        end_time = time.perf_counter()
        together.append((num_points, end_time - begin_time))
        
    ind = np.array(individual)
    tog = np.array(together)
    
    plt.plot(ind[1:, 0], ind[1:, 1], label="Plotted individually")
    plt.plot(tog[1:, 0], tog[1:, 1], label="Plotted together")
    plt.title("Running times for plotting individually vs plotting together")
    #plt.title("Running times for plotting together")
    plt.xlabel("Number of plotted points")
    plt.ylabel("Time in seconds")
    plt.legend()
    plt.show()
    

compare_plot_points(range(100, 1001, 50))

# Comment out generation of individually plotted times and plot only together times
#compare_plot_points(range(100, 10001, 50))
```
