```python
import time
import random
import heapq
```

# Two functions to measure

```python
def sort(array):
    return sorted(array)


def heapsort(array):
    h = []
    for value in array:
        heapq.heappush(h, value)
    return [heapq.heappop(h) for i in range(len(h))]


data = [random.randint(0, 100) for _ in range(10000)]
```

# The naive way

```python
# sort
start = time.perf_counter()
sort(data)
end = time.perf_counter()
print(sort.__name__, "Time Elapsed:", end - start)

# heapsort
start = time.perf_counter()
heapsort(data)
end = time.perf_counter()
print(heapsort.__name__, "Time Elapsed:", end - start)
```

# The decorator way

```python
def timeit(func):
    def wrapped_func(*arg, **kwargs):
        start = time.perf_counter()
        func(*arg, **kwargs)
        end = time.perf_counter()
        
        print(func.__name__, "Time Elapsed:", end - start)
        
    return wrapped_func
```

```python
# sort
@timeit
def sort_2(*arg, **kwargs):
    sort(*arg, **kwargs)

sort_2(data)

# heapsort
@timeit
def heapsort_2(*arg, **kwargs):
    heapsort(*arg, **kwargs)

heapsort_2(data)
```

### What is a decorator?
A function returning another function, usually applied as a function transformation using the @wrapper syntax. Common examples for decorators are classmethod() and staticmethod().

The decorator syntax is merely syntactic sugar, the following two function definitions are semantically equivalent:

~~~
def f(...):
    ...
f = staticmethod(f)

@staticmethod
def f(...):
    ...
~~~
The same concept exists for classes, but is less commonly used there. See the documentation for [function definitions](https://docs.python.org/3.8/reference/compound_stmts.html#function) and [class definitions](https://docs.python.org/3.8/reference/compound_stmts.html#class) for more about decorators.

# The context manager way

```python
class Timer:
    def __init__(self, label=""):
        self.label = label
        
    def __enter__(self):
        self.start = time.perf_counter()
        
        return self

    def __exit__(self, type, value, traceback):
        self.end = time.perf_counter()
        print(self.label, "Time Elapsed:", self.end - self.start)
```

```python
# sort
with Timer(sort.__name__):
    sort(data)

# heapsort
with Timer(heapsort.__name__):
    heapsort(data)
```

# The @contextlib.contextmanager way
https://docs.python.org/3/library/contextlib.html

```python
from contextlib import contextmanager

@contextmanager
def Timer():
    start = time.perf_counter()
    start_end = [start]
    yield start_end
    start_end.append(time.perf_counter())
```

```python
# sort
with Timer() as start_end:
    sort(data)

print(start_end[1] - start_end[0])

# heapsort
with Timer() as start_end:
    heapsort(data)

print(start_end[1] - start_end[0])
```
