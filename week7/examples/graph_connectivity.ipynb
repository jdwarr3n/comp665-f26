```python
"""
Examine graph representations and methods for analyzing graphs
"""

import networkx as nx
import matplotlib.pyplot as plt
```

## Two simple example graphs

```python
# Create a simple graph
grph = nx.Graph()
grph.add_nodes_from([2, 3, 5, 7])
grph.add_edges_from ([(2, 3), (3, 5), (3, 7), (5, 7)])
nx.draw(grph, with_labels=True, node_color="y")
```

```python
# Select a built-in graph
cube = nx.cubical_graph()
print(cube.nodes)
print(cube.edges)
nx.draw(cube, with_labels=True, node_color="y")
```

## Graphs as dicts

```python
# Accessing neighbors of a node - store as a dict

print(grph[3])
print()
print(cube[0])
```

```python
# Convert networkx graph to dict of dicts - standard Python rep
# https://networkx.github.io/documentation/networkx-1.10/reference/convert.html

grph_dict = nx.to_dict_of_dicts(grph)
print(grph_dict)
for node in grph_dict:
    print(node, grph_dict[node])
    
print()
cube_dict = nx.to_dict_of_dicts(cube)
print(cube_dict)
for node in cube_dict:
    print(node, cube_dict[node])
```

## Shortest paths in graphs

An edge connects pair of adjacent nodes.  A **path** in a graph is a sequence of nodes where consecutive nodes are connected by edges. The **length** of this path is the number of edges in the path. 

```python
# Shortest path between two nodes - list of nodes

print(nx.shortest_path(grph, 2, 7))
print(nx.shortest_path(cube, 1, 5))
```

```python
# All pairs shortest paths - returns graph whose edge labels are paths
# Convert to dict of dicts representation

all_shortest = dict(nx.all_pairs_shortest_path(grph))

for node in all_shortest:
    print(node, all_shortest[node])
```

```python
# Extract shortest path from node 2 to node 7 from all_shortest

print(all_shortest[2][7])
```
