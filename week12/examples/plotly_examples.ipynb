# Examples of line plots in plotly

These examples are derived from this [source page](https://plot.ly/python/creating-and-updating-figures/) in plotly. 

```python
import plotly.io as pio
import plotly.graph_objects as go
```

## Figures as dictionaries
`plotly` figure can be defined as a pure Python dictionary. This design makes generating the appropriate JSON for the Javascript version of the figure easy. The dictionary contains two top-level keys:
* `data` - list of traces (graphical objects)
* `layout` - dictionary containing formatting information of the whole figure

```python
# Dictionary defining a line chart in plotly
line_chart = {"data": [{"type": "scatter", "x": [1, 2, 3], "y": [1, 3, 2]}],
              "layout": {"title": "Plot of one line (figure as a dictionary)", 
                         "xaxis_title": "x_axis_label", 
                         "yaxis_title": "y_axis_label"}}

pio.show(line_chart)
```

## Figures as graph objects
The `graph_objects` module allows figures to be manipulated in an object-oriented styles. `Figure()` method creates figure object by either passing a dictionary containing top-level `data` and `layout` keys or setting `data` and `layout` parameters as objects of `plotly.graph_objects` module. Again, `data` is a list of traces and `layout` contains formatting information for the whole figure. 

**Passing a dictionary of `data` and `layout`:**

```python
go_fig1 = go.Figure(dict(data=[dict(type="scatter", 
                                    x=[1, 2, 3], 
                                    y=[1, 3, 2])], 
                         layout=dict(title="Plot of one line (figure as an object)", 
                                     xaxis_title="x_axis_label", 
                                     yaxis_title="y_axis_label")))
go_fig1.show()
```

**Setting `data` and `layout` parameters as objects:**

```python
trace0 = go.Scatter(x=[1, 2, 3], y=[1, 3, 2])

go_fig2 = go.Figure(
    data=[trace0],
    layout=go.Layout(title="Plot of one line (figure as an object)", 
                     xaxis_title="x_axis_label", 
                     yaxis_title="y_axis_label"))
go_fig2.show()
```

We can use the <code>add_trace()</code> method to add more traces to the figure. We can also use the <code>update_layout()</code> method to update the formatting information.

```python
trace1 = go.Scatter(x=[3, 2, 1], y=[1, 1, 2])
go_fig2.add_trace(trace1)
go_fig2.update_layout(title="Plot of two lines")
go_fig2.show()
```

`Figure` object encapsulates the underlying dictionary that contains the figure's information.
