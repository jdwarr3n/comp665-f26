# matplotlib vs. seaborn

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
DATA_PATH = "../data/"
```

```python
iris = pd.read_csv(DATA_PATH + 'iris.csv')
print(iris.head())
```

## Scatter plot 
Scatter plot of petal_length vs. petal_width for different iris classes

**Scatter plot with matplotlib**

```python
# create 3 dataframes, one for each class
setosa = iris[iris['class']=='Setosa']
versicolor = iris[iris['class']=='Versicolor']
virginica = iris[iris['class']=='Virginica']


# create a scatter plot of the sepal_length vs. sepal_width of the three iris classes
fig, ax = plt.subplots()

ax.scatter(setosa['petal_length'], setosa['petal_width'], label='setosa')
ax.scatter(versicolor['petal_length'], versicolor['petal_width'], label='versicolor')
ax.scatter(virginica['petal_length'], virginica['petal_width'], label='virginica')

ax.set_title('Iris Dataset')
ax.set_xlabel('petal_length')
ax.set_ylabel('petal_width')
ax.legend()
```

**Scatter plot with seaborn**

```python
# create a scatter plot of the petal_length vs. petal_width of the three iris classes
fig, ax = plt.subplots()

sns.scatterplot(x='petal_length', y='petal_width', hue='class', data=iris)

ax.set_title('Iris Dataset')
```

## Line chart 
Line chart of all sepal_length, sepal_width, petal_length, and petal_width 

**Line chart with matplotlib**

```python
fig, ax = plt.subplots()

ax.plot(iris['sepal_length'], label='sepal_length')
ax.plot(iris['sepal_width'], label='sepal_width')
ax.plot(iris['petal_length'], label='petal_length')
ax.plot(iris['petal_width'], label='petal_width')

ax.set_title('Iris Dataset')
ax.legend()
```

**Line chart with seaborn**

```python
fig, ax = plt.subplots()

sns.lineplot(data=iris.drop(['class'], axis=1))

ax.set_title('Iris Dataset')
```

## Other high-level plots with seaborn

**Faceting**

```python
fg = sns.FacetGrid(iris, col='class')
fg.map(plt.hist, 'sepal_length')
```

**Plot data with a linear regression model fit**

```python
sns.regplot(data=iris, x='sepal_width', y='petal_width')
```

**Plot data and regression model fits across a FacetGrid**

```python
sns.lmplot(data=iris, x='sepal_width', y='petal_width', col='class')
```

```python
sns.lmplot(data=iris, x='sepal_width', y='petal_width', hue='class')
```
