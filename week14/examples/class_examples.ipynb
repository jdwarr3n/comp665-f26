### A simple example of defining and using a class

```python
import matplotlib.pyplot as plt
```

#### Class definition

```python
class Point:
    """
    Class for creating a 2D point and connecting with other points 
    """
    def __init__(self, x_coord, y_coord):
        """
        Create a point with x and y coordinates
        """
        self._x = x_coord
        self._y = y_coord
        
    def get_x(self):
        """
        Get the x coordinate of the point
        """
        return self._x
    
    def get_y(self):
        """
        Get the y coordinate of the point
        """
        return self._y
    
    def mid_point(self, other):
        """
        Compute the middle point between two points
        """
        mid_x = (self._x + other.get_x()) / 2
        mid_y = (self._y + other.get_y()) / 2
        return (mid_x, mid_y)
    
    def plot_line(self, other):
        """
        Draw a line between the two points and annotating the middle point
        """
        x_values = [self._x, self.mid_point(other)[0], other.get_x()]
        y_values = [self._y, self.mid_point(other)[1], other.get_y()]
        fig, axs = plt.subplots()
        axs.plot(x_values, y_values, 'o-')
        axs.annotate('middle point', xy=self.mid_point(other), xycoords='data')
        
```

#### Class usage

```python
point1 = Point(1, 3)
point2 = Point(4, 2)
point1.plot_line(point2)
```
