# Grammer of Graphics 

Examples below derived from this [tutorial](https://towardsdatascience.com/a-comprehensive-guide-to-the-grammar-of-graphics-for-effective-visualization-of-multi-dimensional-1f92b4ed4149). 

```python
# install plotnine package through "pip install plotnine" in Anaconda Prompt 
from plotnine import *
from plotnine.data import mtcars

import warnings
warnings.filterwarnings("ignore")
```

```python
# data
mtcars.head()
```

```python
(ggplot(mtcars, 
        aes('wt', 'mpg', 
            color='factor(gear)', size='cyl'))
        + geom_point() 
        + theme_bw())
```

```python
(ggplot(mtcars, 
        aes('wt', 'mpg', 
            color='factor(gear)'))
        + geom_point() 
        + facet_wrap('~cyl') 
        + theme_bw())
```

```python
(ggplot(mtcars, 
        aes('wt', 'mpg', 
            color='factor(gear)'))
        + geom_point() 
        + stat_smooth(method='lm') 
        + theme_bw())
```
