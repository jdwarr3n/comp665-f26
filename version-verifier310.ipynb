Python notebook that prints out version names of packages using Comp 665 as well as version in existing Vocareum install of packages. Note that different versions can be a source of problems (but necessarily guarantee them).

```python
import platform
print("Current python version is", platform.python_version())
print("Vocareum python version is 3.10.2")
```

```python
import importlib
import contextlib
import io

def package_info(package_name, voc_version = ""):
    try:
        with contextlib.redirect_stdout(io.StringIO()): # Suppress output during import
            package = importlib.import_module(package_name)
        print("Installed version of", package_name, "is", getattr(package, "__version__"))
    except:
        print(package_name, "not installed")

package_info("networkx", "3.2.1")
```
